```python
import numpy as np
#setting  a random seed so we have the same value on every start
np.random.seed(0)

# three token acting as our demo dataset

x = np.array([[1., 0., 1., 0.],
              [0., 2., 0., 2.],
              [1., 1., 1., 1.]])

# random weight matrices 

Wq = np.random.randn(4, 3)
Wk = np.random.randn(4, 3)
Wv = np.random.randn(4, 3)

def softmax(z):
    e = np.exp(z - z.max(axis=-1, keepdims=True))
    return e / e.sum(axis=-1, keepdims=True)

 # queries
Q = x @ Wq 
  # keys 
K = x @ Wk  
  # values
V = x @ Wv                                  

scores  = Q @ K.T / np.sqrt(K.shape[-1])      # how much each token matches each other token
weights = softmax(scores)                     # each row sums to 1
out     = weights @ V                         # weighted mix of values
#printing output
print("attention weights:\n", weights.round(2))
print("row sums:", weights.sum(axis=1))
print("output:\n", out.round(2))
```
